```python
import os
from dotenv import load_dotenv
from entsoe import EntsoePandasClient
import pandas as pd

load_dotenv()
client = EntsoePandasClient(api_key=os.getenv("ENTSOE_API_KEY"))

start = pd.Timestamp('2024-08-01', tz='Europe/Paris')
end = pd.Timestamp('2026-08-01', tz='Europe/Paris')

fr_generation = client.query_generation('FR', start=start, end=end)

fr_generation.to_csv('data/raw/entsoe_fr_generation_2024-08-01_2026-08-01.csv')
```

```python
last_hourly = gaps[gaps == pd.Timedelta('1 hour')].index.max()
first_15min = gaps[gaps == pd.Timedelta('15 minutes')].index.min()

print("Last hourly timestamp:", last_hourly)
print("First 15-min timestamp:", first_15min)
```
